```python
class Caesar:
    def __init__(self, key: int):
        self.key = key % 26
        
    def encrypt(self, text) -> str:
        res = ""
        for char in text:
            if char.isalpha():
                base = ord('A') if char.isupper() else ord('a')
                res += chr((ord(char) - base + self.key) % 26 + base)
            else:
                res += char
        return res

    def decrypt(self, text) -> str:
        res = ""
        for char in text:
            if char.isalpha():
                base = ord('A') if char.isupper() else ord('a')
                res += chr((ord(char) - base - self.key) % 26 + base)
            else:
                res += char
        return res 
```

```python
# Make a Caesar object
key = 3
caesar = Caesar(key)
```

```python
# Encryption
plaintext = "Hello World"
ciphertext = caesar.encrypt(plaintext)
print(f"Encrypted: {ciphertext}")
```

```python
# Decryption
recovered = caesar.decrypt(ciphertext)
print(f"Encrypted: {recovered}")
assert recovered == plaintext
print("Decryption successfull")
```
